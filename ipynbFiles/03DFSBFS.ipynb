```python
#재귀함수
def recursive_function(i):
    if i==5: #0~4까지
        return
    print(f'{i}->{i+1}')
    recursive_function(i+1)
    print(f'{i} 종료')
```

```python
def factorialforiter(n):
    result=1
    if n<=1:
        return 1
    else:
        for i in range(1,n+1):
            result*=i
        return result
```

```python
def factorialforRecur(n):
    if n<=1:
        return 1
    else:
        return n*factorialforRecur(n-1)
```

```python
#n팩토리얼 구현
#n이 0이나 1이면 결과는 1임
factorialforiter(4)
factorialforRecur(4)
```

```python
def dfs(g,v,visit):
    visit[v]=True
    print(f'{v}노드 방문함')
    for i in g[v]:
        if visit[i]==False:
            dfs(g,i,visit)
```

```python
from collections import deque

def bfs(g,s,visit):
    
    queue = deque([s])
    
    visit[s] = True
    
    while queue:
        v = queue.popleft()
        print(f'{v} 나가리')
        
        for i in g[v]:
            if visit[i]==False:
                queue.append(i)
                visit[i]=True  
```

```python
#n,m = map(int, input().split()) #세로, 가로
#order = [input() for i in range(n)]
n,m = 4,5
order=['00110','00011','11111','00000']

graph = [[0]*m for i in range(n)]
for i in range(n):
    for j in range(m):
        if order[i][j]=='1':
            graph[i][j]=1
print(graph)

def dfs(x,y):
    if x<=-1 or x>=n or y<=-1 or y>=m: #해당 위치가 맵에서 벗어나는 지 확인
        return False
    
    if graph[x][y]==0: #얼음이 뚤려있는 곳인지 확인
        graph[x][y]=1 #얼음칸 막음 -> 사방팔방 파고들어서 뚤린곳 다 막음
        dfs(x-1,y)
        dfs(x+1,y)
        dfs(x,y-1)
        dfs(x,y+1)
        return True # 끝
    else: #이미 방문 했던 곳
        return False

result = 0
for i in range(n): # 행
    for j in range(m): # 열
        if dfs(i,j) == True:
            result+=1

print(result)
```

```python
n,m = 6,5
order = ['11011',
         '11101',
         '11111',
         '00001',
         '11111',
         '11111'
        ]
graph = [[0]*m for i in range(n)]
for i in range(n):
    for j in range(m):
        if order[i][j]=='1':
            graph[i][j]=1
print(graph)

def dfs(y,x):
    if x<=-1 or x>=m or y<=-1 or y>=n:
        return 0
    print(f'현재 위치 x:{x} y:{y}')
    graph[y][x]=0
    
    if x==m-1 and y==n-1:
        print('조건들어옴')
        return 1
    
    if -1<(x+1)<m and graph[y][x+1]!=0:#장애물이 없으면
        return dfs(y,x+1)+1
    if -1<(y+1)<n and graph[y+1][x]!=0:#장애물이 없으면
        return dfs(y+1,x)+1
    if -1<(x-1)<m and graph[y][x-1]!=0:#장애물이 없으면
        return dfs(y,x-1)+1
    if -1<(y-1)<n and graph[y-1][x]!=0:#장애물이 없으면
        return dfs(y-1,x)+1
print(dfs(0,0))
```

```python
n,m = 5,6
order = ['101010',
         '111111',
         '000001',
         '111111',
         '111111',
        ]
graph = [[0]*m for i in range(n)]
for i in range(n):
    for j in range(m):
        if order[i][j]=='1':
            graph[i][j]=1
print(graph)

def bfs(graph):
    queue = deque()
    queue.append((0,0)) # 여러개를 저장할 수 있지만 한개씩 처리할 수도 있음
    
    dxy=[(1,0),(-1,0),(0,-1),(0,1)]
    
    while queue:
        print(f'현재 큐 {queue}')
        x,y = queue.popleft()
        print(f'검색빠진 큐 {queue}')
        for k in dxy:
            dx = x+k[0]
            dy = y+k[1]
            if dx<=-1 or dx>=m or dy<=-1 or dy>=n:
                continue
            if graph[dy][dx]==0:
                continue
            if graph[dy][dx]==1: #처음 위치에 도달했을때(최단거리니까)
                graph[dy][dx]=graph[y][x]+1
                queue.append((dx,dy))
    return graph[n-1][m-1]

# 동서남북
print(bfs(graph))
```
