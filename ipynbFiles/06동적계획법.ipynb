### 탑다운 다이나믹 프로그래밍
- 피보나치 수열


```python
d=[0]*100
def fibo(x):
    if x==1 or x==2:
        return 1
    if d[x]!=0:
        return d[x]
    d[x]=fibo(x-1)+fibo(x-2)
    return d[x]

print(fibo(999))
```

### 보텀업 다이나믹 프로그래밍
- 피보나치 수열

```python
d=[0]*100

d[1]=1
d[2]=1
n=99

for i in range(3,n+1):
    d[i]=d[i-1]+d[i-2]

print(d[n])
```

```python
n=4
mylist=[1,3,1,5]

d=[0]*n

d[0]=mylist[0]
d[1]=max(mylist[0],mylist[1])

idx=2
while idx<len(mylist):
    d[idx]=max(d[idx-1],mylist[idx]+d[idx-2])   
    idx+=1

print(d[n-1])
```

```python
array=[2,3,5]
m=7
n=3

d=[10001]*(m+1) #한번도 접근 안한 거 구분하려고 10001

d[0]=0 #기준값에 대해 0+1 연산하려고

for i in range(n): #기준값을 한 과정에 한 번씩
    for idx in range(array[i],m+1): #기준값부터 검색값까지 모두
        if d[idx-array[i]]!=10001:
            d[idx]=min(d[idx-array[i]]+1, d[idx]) #d[idx]를 보는 이유 기준 값 차례로 적용하면서 수정된 내용도 보는거

if d[m-1]==10001:
    print(-1)
else:
    print(d[m-1])
```

```python
'''
n=3
m=4
graph='1 3 3 2 2 1 4 1 0 6 4 7'
'''
n=4
m=4
graph='1 3 1 5 2 2 4 1 5 0 2 3 0 6 1 2'

graph=graph.split(' ')
array = [[0]*m for _ in range(n)]
idx=0
for i in range(n):
    for j in range(m):
        array[i][j]=int(graph[idx])
        idx+=1
print(array)

for x in range(1,m):
    for dy in range(n):
        ranged=[i for i in range(dy-1,dy+2)]
        tmp=[]
        for j in ranged:
            if 0<=j<n:
                tmp.append(array[j][x-1])
        array[dy][x]=max(tmp)+array[dy][x]

tmp=[]
for idx in range(n):
    tmp.append(array[idx][m-1])
print(max(tmp))
```

### 요약
- 1. 내림차순으로 정리하는 문제 -> 오름차순으로 뒤집기
- 2. dp 테이블에 1로 채우기 개수 카운트
- 3. 현재 인덱스 기준까지 앞에 원소와 하나씩 비교하기 -> 여기 까지 계산했을 때 가능한 배열의 길이가 얼마만큼인지 확인하는 과정
- 4. 오름차순이니까 앞원소보다 뒤원소가 커야함 -> 배열 길이가 증가하니까 dp 원소도 증가해야함
- 5. 가능한 배열의 길이를 구하고 전체원소에서 뺀 값이 최소 개수
