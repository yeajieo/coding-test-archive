```python
mylist=[7,5,9,0,3,1,6,2,4,8]

print(mylist)
for _ in range(1,len(mylist)):    
    target=mylist[_]
    listCompared=mylist[:_]
    for idx in range(len(listCompared)-1,-1,-1):
        if target<listCompared[idx]:
            a=mylist.index(listCompared[idx])
            b=mylist.index(target)
            mylist[a],mylist[b]=mylist[b],mylist[a]
        else:
            break
print(mylist)
```

```python
def quickSort(mylist): 
    
    if len(mylist)<=1:
        return mylist
    
    pivot=mylist[len(mylist)//2]

    small=[]
    mid=[]
    big=[]
    for _ in mylist:
        if pivot>_: small.append(_)
        elif pivot<_: big.append(_)
        else: mid.append(_)           
    
    return quickSort(small)+mid+quickSort(big)
```

```python
mylist=[3,9,4,7,5,0,1,6,8,2]
quickSort(mylist)
```

```python
def countingSort(mylist):
    array = [0]*(max(mylist)+1)
    for num in mylist:
        array[num]+=1
    result=[]
    for idx,count in enumerate(array):
        for _ in range(count):
            result.append(idx)
    return result
```

```python
mylist=[7,5,9,0,3,1,6,2,9,1,4,8,0,5,2]
countingSort(mylist)
```

```python
def quickSort(mylist):
    if len(mylist)<=1:
        return mylist
    pivot=mylist[len(mylist)//2]
    s=[]
    m=[]
    l=[]
    for num in mylist:
        if pivot>num: s.append(num)
        elif pivot<num: l.append(num)
        else: m.append(num)
    return quickSort(s)+m+quickSort(l)
```
