```python
n,k = map(int, input().split())

result=0 #총 연산 횟수

while True:
    
    # 기준 1번 수행 - n을 k로 나눴을 때 나머지를 먼저 생각하는 과정
    target = (n//k)*k 
    result += (n-target) #n을 k로 나눈 나머지 생각 - 먼저 카운트 
    n=target # n-나머지 값 == k로 나누어 떨어질 수만 남김
    
    if n<k: #현재 값이 나누는 수보다 작으면
        break
    
    result+=1 # 횟수 추가
    n//=k #기준 2번 수행
    
```

```python
# 내 코드 - 틀린 풀이
# 내 풀이: 1. 오름차순 정렬하기 2. 뒤에서 부터 그룹짓기
# 정답: 1. 오름차순 정렬하기 2. 앞에서 부터 그룹짓기

group = list(map(int, input().split())) # 개수 지정 X, 공백 구분
#group = [2,3,1,2,2]
group = sorted(group)
print(group)

count = 0

while True:
    if len(group)!=0:
        target = group[-1]
        print(target)
        
        if len(group) >= target:
            group = group[target-1:-1]
            print(group)
            count+=1
        else:
            break
    else:
        break
print(count)
```

```python
def getDirection(curX,curY,direct):
    # 동북서남
    dx = [0,-1,0,1]
    dy = [1,0,-1,0]
    if direct=='R':#동
        curX+=dx[0]
        curY+=dy[0]
    elif direct == 'U':#북
        curX+=dx[1]
        curY+=dy[1]
    elif direct == 'L':#서
        curX+=dx[2]
        curY+=dy[2]
    else: #남
        curX+=dx[3]
        curY+=dy[3]
    return curX, curY
```

```python
#n=int(input())
#data=list(map(str, input().split()))

n=5
data=['R','R','R','U','D','D']

x=1
y=1

for i in data:
    print(i)
    tmpX,tmpY=getDirection(x,y,i)
    if 1<=tmpX<=n and 1<=tmpY<=n:
        x,y=tmpX,tmpY
    else:
        continue

print(x,y)
```
