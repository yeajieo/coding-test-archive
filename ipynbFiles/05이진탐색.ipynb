### 예제. 값이 특정 범위에 속하는 데이터 개수 구하기

```python
mylist=[1,2,2,3,3,3,3,4,4,5]
num=[5]

def countNum(array,n):
    if len(n)==1:
        return bisect_right(array,n[0])-bisect_left(array,n[0])
    else:
        return bisect_right(array,n[1])-bisect_left(array,n[0])
print(countNum(mylist,num))
        
```
