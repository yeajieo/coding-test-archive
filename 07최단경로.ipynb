### 최소 힙
- 오름차순

```python
import heapq

def heapsort(iterable):
    h=[]
    result=[]
    
    # 이진 트리 상태로 저장
    for value in iterable:
        heapq.heappush(h,value)
        print(h)
    # 트리 특성상 작은 값부터 출력
    for i in range(len(h)):
        result.append(heapq.heappop(h))
    return result

result=heapsort([1,3,5,7,9,2,4,6,8,0])
print(result)
```

### 최대 힙

```python
import heapq

def heapsort(iterable):
    h=[]
    result=[]
    
    # 이진 트리 상태로 저장
    for value in iterable:
        heapq.heappush(h,-value)
        print(h)
    # 트리 특성상 작은 값부터 출력
    for i in range(len(h)):
        result.append(-heapq.heappop(h))
    return result

result=heapsort([1,3,5,7,9,2,4,6,8,0])
print(result)
```

```python
def heapsort(mylist): # 가중치가 같으면, 노드가 작은 순 먼저
    h=[]
    sortedMylist=[]
    # 이진 트리 상태로 저장
    for d,n in mylist:
        heapq.heappush(h,(d,n))
    # 트리 특성상 작은 값부터 출력
    for i in range(len(h)):
        sortedMylist.append(heapq.heappop(h))
    return sortedMylist
```

```python
def makeArray(graph,n):
    array=[]
    for node in graph[n]:
        array.append((node[1],node[0]))
    return array
```

```python
import heapq
from collections import deque

INF = int(1e9)
n,m= 6,11
start=1
distance=[INF]*(n+1)

graph=[[],
        [(2,2),(3,5),(4,1)],
        [(3,3),(4,2)],
        [(2,3),(6,5)],
        [(3,3),(5,1)],
        [(3,1),(6,2)],
        []
    ]

#초기단계
distance[start]=0
queue=deque()
queue.append((distance[start],start))
d,n=queue.popleft()

count=0
while 1:
    if d>distance[n]:
        if len(queue)!=0:
            d,n=queue.popleft()
            continue
        else:
            break  
    nextList=makeArray(graph,n)
    total=[]
    tmp=[]
    for i in nextList:
        num=distance[n]+i[0]
        if distance[i[1]]>num:
            distance[i[1]]=num
            tmp.append((num,i[1]))
    total=list(queue)+tmp
    sortedTotal = heapsort(total)
    queue=deque(sortedTotal)
    d,n=queue.popleft()

print(distance[1:])
```

### 정석 코드

```python
def dijkstra(start):
    q = []
    heapq.heappush(q, (0, start))  # 시작노드 정보 우선순위 큐에 삽입
    distance[start] = 0            # 시작노드->시작노드 거리 기록
    while q:
        dist, node = heapq.heappop(q)
        # 큐에서 뽑아낸 거리가 이미 갱신된 거리보다 클 경우(=방문한 셈) 무시
        if distance[node] < dist:
            continue
        # 큐에서 뽑아낸 노드와 연결된 인접노드들 탐색
        for next in graph[node]:
            cost = distance[node] + next[1]   # 시작->node거리 + node->node의인접노드 거리
            if cost < distance[next[0]]:      # cost < 시작->node의인접노드 거리
                distance[next[0]] = cost
                heapq.heappush(q, (cost, next[0]))
```
